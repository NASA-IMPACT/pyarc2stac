```python
from pyarc2stac.ArcReader import ArcReader
```

```python
image_server_url = 'https://gis.earthdata.nasa.gov/UAT/rest/services/disasters_test/lis_vsm_percentile_10cm/ImageServer'
```

```python
reader = ArcReader(
    server_url = image_server_url
)
```

```python
collection = reader.generate_stac()
```

```python
collection.to_dict()
```

```python
reader.save_collection_to_json()
```
